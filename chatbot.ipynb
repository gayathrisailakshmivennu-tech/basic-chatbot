```python
def chatbot():
    print("🤖 Chatbot: Hello! I am a simple chatbot.")
    print("🤖 Chatbot: Type 'bye' to exit.")

    while True:
        user_input = input("You: ").lower()

        if user_input == "hello" or user_input == "hi":
            print("Chatbot: Hi! Nice to meet you.")

        elif user_input == "how are you":
            print("Chatbot: I'm fine, thanks! How are you?")

        elif user_input == "what is your name":
            print("Chatbot: My name is Python Chatbot.")

        elif user_input == "what can you do":
            print("Chatbot: I can have a simple conversation with you.")
            print("Chatbot: I can answer simple questions.")wht

        elif user_input == "tinnava":
            print("chatbot: tinna nuvvu tinnava")

        elif user_input == "thank you" or user_input == "thanks":
            print("Chatbot: You're welcome!")

        elif user_input == "bye":
            print("Chatbot: Goodbye! Have a nice day!")
            break

        else:
            print("Chatbot: Sorry, I don't understand that.")


chatbot()
```
